```python


def read_files_from_directory(directory):
    data_dict = {}
    skip_filename = "curnpt.ipynb"
    
    for filename in os.listdir(directory):
        if filename == skip_filename:
            continue  # Skip the specified file
        
        file_path = os.path.join(directory, filename)
        
        if os.path.isfile(file_path):  # Ensure it's a file
            try:
                data = np.genfromtxt(file_path, dtype=float,delimiter=" ",skip_header=1)
                data_dict[filename] = data  # Include extension in key
            except Exception as e:
                print(f"Error reading {filename}: {e}")
                
    return data_dict

# Example usage
directory_path = "nvts"
files_data = read_files_from_directory(directory_path)
#print(files_data)
```
